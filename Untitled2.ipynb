```python
# prompt: crie um código que permite 3 opções de calculos: cilindro, esfera e cubo
import math

def calcular_cilindro():
  raio = float(input("Digite o raio da base do cilindro: "))
  altura = float(input("Digite a altura do cilindro: "))
  area = 2 * math.pi * raio * (raio + altura)
  volume = math.pi * raio**2 * altura
  print("Área da superfície do cilindro:", area)
  print("Volume do cilindro:", volume)

def calcular_esfera():
  raio = float(input("Digite o raio da esfera: "))
  area = 4 * math.pi * raio**2
  volume = (4/3) * math.pi * raio**3
  print("Área da superfície da esfera:", area)
  print("Volume da esfera:", volume)

def calcular_cubo():
  lado = float(input("Digite o lado do cubo: "))
  area = 6 * lado**2
  volume = lado**3
  print("Área da superfície do cubo:", area)
  print("Volume do cubo:", volume)

while True:
  print("\nCalculadora Geométrica")
  print("1. Cilindro")
  print("2. Esfera")
  print("3. Cubo")
  print("4. Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == '1':
    calcular_cilindro()
  elif opcao == '2':
    calcular_esfera()
  elif opcao == '3':
    calcular_cubo()
  elif opcao == '4':
    break
  else:
    print("Opção inválida!")
```

```python
Lista=[4.5, 5.6, 7.7, 3.5, 8.9, 9.0, 4.5]
soma=0
for nota in Lista:
  soma=soma+nota
media=soma/len(Lista)
print("A média é: ", '{:.4f}'.format(media))
```

```python
#Criação da função para calculo das médias para uma lista
def CalcMedia(lista):
  soma=0
  for nota in lista:
    soma=soma+nota
  return soma/len(lista)
```

```python
Lista1=[4.5, 5.6, 7.7, 3.5, 8.9, 9.0, 4.5]
media=CalcMedia(Lista1)
print("A média é: ", '{:.4f}'.format(media))
```
